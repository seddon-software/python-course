## Q5
Use Pandas to read the file "wtk_site_metadata.csv"
Create a new dataframe containing rows with state equal to 'Rhode Island'.

## Q7
Use Pandas to read the file "wtk_site_metadata.csv"
Create the same dataframe 'Rhode Island' as in the previous exercise.
Now add an extra column called 'power' that is computed as the product of the 
columns 'capacity_factor' and 'wind_speed'.  Then print the last 20 rows of the new dataframe.

```python
import pandas as pd

pd.set_option('display.precision', 2)
pd.set_option('display.width', None)
pd.set_option('display.max_rows', None)

df = pd.read_csv("wtk_site_metadata.csv")
df['power'] = df.apply(lambda row:row.capacity_factor*row.wind_speed, raw=False, axis=1)
print(df[['capacity_factor','wind_speed','power']][:20])
```

## Q8
Use Pandas to read the file "wtk_site_metadata.csv"
Work with the dataframe 'Rhode Island' as in the previous exercise with the
extra column called 'power'.
Reorder the columns such that the first three columns are 'capacity_factor',
'wind_speed' and 'power' and print the last 20 rows of the dataframe.

```python
import pandas as pd

pd.set_option('display.precision', 2)
pd.set_option('display.width', None)
pd.set_option('display.max_rows', None)

df = pd.read_csv("wtk_site_metadata.csv")
df['power'] = df.apply(lambda row:row.capacity_factor*row.wind_speed, raw=False, axis=1)

c = df.columns
df = df[[c[8],c[9],c[-1],*c[:8],*c[10:-1]]]
print(df.tail(20))
```

## Q9
Use Pandas to read the file "wtk_site_metadata.csv"
Work with the dataframe 'Rhode Island' as in the previous exercise with the
extra column called 'power' and the column headings resorted.
Now sort the dataframe in descending order of 'power' and print the last 20 rows.

```python
import pandas as pd

pd.set_option('display.precision', 2)
pd.set_option('display.width', None)
pd.set_option('display.max_rows', None)

df = pd.read_csv("wtk_site_metadata.csv")
df['power'] = df.apply(lambda row:row.capacity_factor*row.wind_speed, raw=False, axis=1)

c = df.columns
df = df[[c[8],c[9],c[-1],*c[:8],*c[10:-1]]]

df.sort_values('power', axis=0, ascending=False, inplace=True)
print(df.tail(20))
```

## Q10
Use Pandas to read the file "wtk_site_metadata.csv"
Work with the dataframe 'Rhode Island' as in the previous exercise with the
extra column called 'power' and the column headings resorted.
Create a new sorted dataframe with rows from the 'Rhode Island' dataframe with
a 'power' greater than 7 and print the whole dataframe. 

```python
import pandas as pd

pd.set_option('display.precision', 2)
pd.set_option('display.width', None)
pd.set_option('display.max_rows', None)

df = pd.read_csv("wtk_site_metadata.csv")
df['power'] = df.apply(lambda row:row.capacity_factor*row.wind_speed, raw=False, axis=1)

c = df.columns
df = df[[c[8],c[9],c[-1],*c[:8],*c[10:-1]]]

df = df[df['power'] > 7]
df.sort_values('power', axis=0, ascending=False, inplace=True)
print(df)
```
